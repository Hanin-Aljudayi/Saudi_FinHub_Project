```python
import requests
import pandas as pd
```

```python
url = "https://api.frankfurter.dev/v2/rates?base=SAR&from=2013-01-01&quotes=USD,EUR,GBP,CNY,AED"

try:
    response = requests.get(url)
    
    # Check for HTTP errors automatically
    response.raise_for_status() 
    
    df = pd.DataFrame(response.json())
    output_dir = r"/home/dania/SaudiFinHub/Saudi_FinHub_Project/data/raw"  # Use 'r' before the string to avoid syntax issues with backslashes
    file_namej = "raw.json"
    file_name = "raw.csv"
    df.to_json(f"{output_dir}/{file_namej}", orient='records', lines=True)  # Save as JSON
    df.to_csv(f"{output_dir}/{file_name}", index=False)
    print("Successfully retrieved data!")
    print(df.head())  # Display the first few rows of the DataFrame
    
except requests.exceptions.HTTPError as err:
    print(f"HTTP error occurred: {err}")
except Exception as err:
    print(f"An error occurred: {err}")
```
