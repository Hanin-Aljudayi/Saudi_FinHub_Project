## Source Definition

- **Source:** Tasilab API
- **Data:** Saudi Stock Market Historical Data
- **Endpoint:** `/v1/historical/{symbol}`
- **Authentication:** API Key using `X-API-Key`
- **Rate Limit:** 120 API requests/minute (Free tier)
- **License/Terms:** Data usage is subject to Tasilab Terms of Use and the terms of its underlying market-data providers.

```python
import os
import json
from datetime import date
from pathlib import Path

from tasilab import Tasilab

tasi = Tasilab(api_key=os.environ["TASILAB_API_KEY"])
```

```python
data = tasi.get_historical(
    "2080",
    "2013-01-01",
    "2026-03-31"
)

print("Records:", data["count"])
print("Period:", data["start"], "to", data["end"])
```

```python
# Extract historical stock data for all selected companies

symbols = [
    "1120",  # Al Rajhi Bank
    "2080",  # GASCO
    "2010",  # SABIC
    "2310",  # Sipchem
    "7010",  # stc
    "7020",  # Mobily
    "1211",  # Ma'aden
    "2280",  # Almarai
    "8210",  # Bupa Arabia
    "4003",  # Extra
    "3060",  # Yanbu Cement
    "3030",  # Saudi Cement
    "4190",  # Jarir Marketing
    "5110",  # Saudi Electricity
    "4030"   # Bahri
]
raw_folder = Path("../data/raw")
raw_folder.mkdir(parents=True, exist_ok=True)

extraction_date = date.today().strftime("%Y%m%d")

for symbol in symbols:
    try:
        data = tasi.get_historical(
            symbol,
            "2013-01-01",
            "2026-03-31"
        )

        file_path = raw_folder / f"tasilab_{symbol}_{extraction_date}.json"

        with open(file_path, "w", encoding="utf-8") as f:
            json.dump(data, f, ensure_ascii=False, indent=2)

        print(
            f"Saved: {file_path} | "
            f"Records: {data['count']} | "
            f"Period: {data['start']} to {data['end']}"
        )

    except Exception as e:
        print(f"Error extracting {symbol}: {e}")
```

### Extraction Notes
- Historical daily stock data was extracted from 2013-01-01 to 2026-03-31
- The original API responses were saved as separate JSON files in `data/raw/`.
- The raw responses were saved without modifying their original structure.
- Extraction errors are handled in the code and reported for the affected stock symbol.
- Tasilab Free Tier allows up to 120 API requests per minute
