## Saudi FinHub — Economic Indicators

This notebook handles the extraction of the economic indicators used in the project:
- GDP
- Inflation
- Unemployment
The purpose of this task is to collect the source data and save the raw data in `data/raw/` without modifying it.

```python
# Import the libraries needed for data extraction
from pathlib import Path   # Used to work with project folders and file paths
```

```python
# Define the folder where the original/raw data will be stored

RAW_DATA_PATH = Path("../data/raw")

# Create the folder if it does not already exist
RAW_DATA_PATH.mkdir(parents=True, exist_ok=True)


print("Raw data folder:", RAW_DATA_PATH)
```

## GDP Source

- **Source Name:** General Authority for Statistics (GASTAT)
- **Dataset:** Gross Domestic Product at Current Prices by Main ISIC
- **Source File:** Gross domestic product (GDP) at current price by Main ISIC_data.csv
- **Data Format:** CSV
- **Frequency:** Quarterly
- **Authentication:** Not required

```python
# Define the raw CSV source files
gdp_raw_file = RAW_DATA_PATH / "Gross domestic product (GDP) at current price by Main ISIC_data.csv"
inflation_raw_file = RAW_DATA_PATH / "Annual Change in Consumer Prices (National, %)_data.csv"
unemployment_raw_file = RAW_DATA_PATH / "Unemployment_rate_by_nationality_and_sex_2013_2026.csv"

# Check that all raw source files exist
for file in [gdp_raw_file, inflation_raw_file, unemployment_raw_file]:
    if file.exists():
        print(f"Found: {file.name}")
    else:
        print(f"Not found: {file.name}")
```
