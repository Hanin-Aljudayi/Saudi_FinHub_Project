# Task 1 — TASI Data Extraction

## Source Definition

- **Source:** Saudi Exchange Historical Market Data
- **Dataset:** Tadawul All Share Index (TASI)
- **Data Type:** Daily Market Index Data
- **Period:** 2014-01-01 to 2026-09-20
- **Frequency:** Daily

## Extraction Objective

Extract the historical TASI daily market data and preserve the original source values without cleaning or transformation.

## Output

The extracted raw dataset will be saved to:

`data/raw/tasi_raw.csv`

The raw dataset will be used as the input for the profiling and cleaning stage in `02_profile_clean.ipynb`.

```python
# Step 1 — Extract the source TASI CSV into the RAW layer
# The source file is copied without modification to preserve the original data.

from pathlib import Path
import shutil

# Source file downloaded from the data provider
source_path = Path(r"C:\Users\aldos\Desktop\tasi_historical.csv")

# Project RAW directory
raw_folder = Path("../data/raw")
raw_folder.mkdir(parents=True, exist_ok=True)

# RAW output file
raw_path = raw_folder / "tasi_raw.csv"

# Validate source file exists
if not source_path.exists():
    raise FileNotFoundError(
        f"Source file was not found: {source_path}"
    )

# Copy the original source file without modifying its contents
shutil.copy2(source_path, raw_path)

print("===== RAW EXTRACTION =====")
print("Source file:", source_path)
print("RAW file:", raw_path.resolve())
print("File size (KB):", round(raw_path.stat().st_size / 1024, 2))
print("Status: RAW file copied successfully.")
```

## Step 2 — Validate Extracted RAW Data

The extracted TASI dataset is loaded from the RAW layer to verify that the extraction was completed successfully.

This step validates:
- File accessibility
- Dataset shape
- Column names
- Sample records

No cleaning or transformation is applied in the extraction stage.

```python
# Step 2 — Load and validate the extracted RAW dataset

import pandas as pd

raw_path = "../data/raw/tasi_raw.csv"

raw_df = pd.read_csv(raw_path)

print("===== RAW DATA VALIDATION =====")

print("\nShape:")
print(raw_df.shape)

print("\nColumns:")
print(raw_df.columns.tolist())

print("\nFirst 5 rows:")
display(raw_df.head())

print("\nLast 5 rows:")
display(raw_df.tail())

print("\nDate coverage:")
print("First date:", raw_df.iloc[:, 0].iloc[0])
print("Last date:", raw_df.iloc[:, 0].iloc[-1])
```

# Stock Market Data Reference

## Source Definition

- **Source:** Saudi Exchange Historical Market Data
- **Dataset:** Tadawul All Share Index (TASI) Historical Data
- **Data Type:** Daily Market Index Data
- **Historical Period:** 2014-01-01 to 2026-09-20
- **Market:** Saudi Exchange (Tadawul)
- **Frequency:** Daily

## Extracted Fields

The extracted TASI dataset contains the following daily market indicators:

- Date
- Close
- Open
- High
- Low
- Volume
- Change%

## Extraction Notes

- Historical TASI market data was collected for the period from 2014-01-01 to 2026-09-20.
- The extracted data represents the Tadawul All Share Index (TASI), rather than individual company-level stock data.
- The original extracted values were preserved before the cleaning and validation stages.
- Date values contained hidden right-to-left characters and were normalized during the cleaning stage.
- Price fields were converted from formatted text to numeric values.
- Volume values containing `M` and `B` units were converted to numeric values.
- Change% values were converted from percentage-formatted text to numeric values.
- Final column names were standardized to English for downstream processing and analytics.
